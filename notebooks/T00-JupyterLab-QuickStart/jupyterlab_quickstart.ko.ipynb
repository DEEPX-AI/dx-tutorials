## 1. Jupyter Notebook과 JupyterLab이란?

### 1.1 Jupyter Notebook
Jupyter Notebook은 다음을 하나로 묶은 파일(`.ipynb`)입니다.
- Python 코드
- 텍스트 문서
- 출력 결과
- 그래프와 로그

다음 분야에서 널리 사용됩니다.
- 데이터 과학
- AI 학습
- 연구
- 교육

### 1.2 JupyterLab
JupyterLab은 노트북이 실행되는 웹 기반 환경입니다.

다음을 할 수 있습니다.
- 여러 노트북을 동시에 열기
- 파일 관리
- 터미널 열기
- 코드를 대화형으로 실행

간단히 정리하면 다음과 같습니다.
- **Notebook = 문서**
- **JupyterLab = 작업 공간**

## 3. 셀 소개 (Markdown 셀 & 코드 셀)

### 3.3 셀 생성 및 `Code cell`에서 `Markdown cell`로 변환
Jupyter에서 **새 Markdown 셀을 만들려면** 툴바의 `+` 버튼을 클릭합니다. Jupyter의 새 셀은 기본적으로 모두 코드 셀로 시작하므로, Markdown 셀로 인식되고 렌더링되려면 셀 형식을 바꿔야 합니다. 먼저 커서로 셀을 클릭해 활성화합니다. 그다음 툴바에서 “Code”라고 표시된 드롭다운 상자(⏭ 버튼 옆에 있습니다)를 클릭해 “Code”를 “Markdown”으로 변경합니다.
![](https://datasciencebook.ca/_main_files/figure-html/convert-to-markdown-cell-1.png)

## 4. 셀 실행과 규칙

코드는 **커널(Kernel)**이라는 프로세스 안에서 실행됩니다.

중요한 규칙:

1. 모든 셀은 같은 메모리를 공유합니다.
2. 실행 순서가 중요합니다.
3. 결과가 이상하면 커널을 재시작하세요.
4. 깔끔한 결과를 얻으려면 위에서 아래로 순서대로 실행하세요.


**코드 셀을 개별적으로 실행하려면** 먼저 셀을 활성화해야 합니다. 커서로 셀을 클릭하면 됩니다. 셀이 활성화되면 Jupyter는 셀 왼쪽에 파란색 사각형을 표시합니다. 셀이 활성화된 뒤에는 툴바의 실행(▶) 버튼을 누르거나 키보드 단축키 `Shift + Enter`로 셀을 실행할 수 있습니다.

![](https://datasciencebook.ca/_main_files/figure-html/activate-and-run-button-1.png)

> **다음 셀에서 예상되는 오류:** `a`가 아직 정의되지 않았으므로 `print(a)`를 먼저 실행하면 `NameError`가 발생합니다. 실행 순서를 보여 주기 위한 의도적인 시연입니다. 오류를 확인한 뒤 `a`를 정의하는 다음 셀을 실행하고, 그다음 두 번째 `print(a)` 셀을 실행하세요.

```python
# Expected: NameError, because a is defined in the next cell.
print(a)
```

```python
# Example: Execution order matters
a = 20
b = 4
a / b
```

```python
print(a)
```

## 6. 실행 중인 셀 중단

코드 셀이 실행 중일 때는 셀 왼쪽에 `*`가 표시됩니다. 셀이 현재 실행 중이라는 뜻입니다.
> [*]

셀 실행이 끝나면 `*`가 숫자로 바뀝니다. 이 숫자는 실행 순서를 나타냅니다.

> [1]</br>
> [2]</br>
> [3]</br>

셀이 너무 오래 실행되면(예: 무한 루프) 툴바의 중지 버튼(■)을 클릭할 수 있습니다.

![](https://media.geeksforgeeks.org/wp-content/uploads/20231010191734/Screenshot-from-2023-10-10-18-55-07.png)

```python
import time

for i in range(10):
    print(f"Running... {i+1}/10")
    time.sleep(3)

print("Done.")
```

## 8. 유용한 키보드 단축키

실행:
- `Shift + Enter` → 셀을 실행하고 다음 셀로 이동 (가장 많이 사용)
- `Ctrl + Enter` → 셀을 실행하고 현재 셀에 머무름

셀 제어:
- `A` → 위에 삽입
- `B` → 아래에 삽입
- `D, D` → 삭제
- `M` → Markdown으로 변경
- `Y` → 코드로 변경

모드:
- `Esc` → 명령 모드
- `Enter` 또는 `Mouse L btn double-click` → 편집 모드
