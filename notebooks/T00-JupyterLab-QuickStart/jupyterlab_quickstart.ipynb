## 1. What is Jupyter Notebook and JupyterLab?

### Jupyter Notebook
A Jupyter Notebook is a file (`.ipynb`) that combines:
- Python code

- Output results

It is widely used for:
- Data science

- Education

### JupyterLab
JupyterLab is the web-based environment where notebooks run.

It allows you to:
- Open many notebooks at once
- Manage files
- Open terminals
- Run code interactively

Simple idea:
- **Notebook = the document**
- **JupyterLab = the workspace**

### Create a cell and convert from `Code cell` to `Markdown cell`
**To create a new Markdown cell** in Jupyter, click the `+` button in the toolbar. By default, all new cells in Jupyter start as code cells, so the cell format needs to be changed to be recognized and rendered as a Markdown cell. To do this, click on the cell with your cursor to ensure it is activated. Then click on the drop-down box on the toolbar that says “Code” (it is next to the ⏭ button), and change it from “Code” to “Markdown”:
![](https://datasciencebook.ca/_main_files/figure-html/convert-to-markdown-cell-1.png)

## 4. Cell Execution and Rules

Code runs inside a process called the **Kernel**.

Important rules:

1. All cells share the same memory.
2. Execution order matters.
3. Restart kernel if results are strange.
4. Run from top to bottom for clean results.


**To run a code cell** independently, the cell needs to first be activated. This is done by clicking on it with the cursor. Jupyter will indicate a cell has been activated by highlighting it with a blue rectangle to its left. After the cell has been activated, the cell can be run by either pressing the Run (▶) button in the toolbar, or by using the keyboard shortcut `Shift + Enter`.

![](https://datasciencebook.ca/_main_files/figure-html/activate-and-run-button-1.png)

> **Expected error in the next cell:** `a` has not been defined yet, so running `print(a)` first raises `NameError`. This is an intentional demonstration of execution order. After observing the error, run the following cell that defines `a`, and then run the second `print(a)` cell.


```python
# Expected: NameError, because a is defined in the next cell.
print(a)
```

```python
# Example: Execution order matters
a = 20
b = 4
a / b
```

```python
print(a)
```

## 6. Interrupt a running cell

When a code cell is executing, you will see a `*` on the left side of the cell. This means the cell is currently running.
> [*]

After the cell finishes, the `*` changes to a number. The number shows the execution order.

> [1]</br>
> [2]</br>
> [3]</br>

If a cell runs too long (for example, an infinite loop), you can click the stop button (■) in the toolbar:

![](https://media.geeksforgeeks.org/wp-content/uploads/20231010191734/Screenshot-from-2023-10-10-18-55-07.png)

```python
import time

for i in range(10):
    print(f"Running... {i+1}/10")
    time.sleep(3)

print("Done.")
```

## 8. Useful Keyboard Shortcuts

Run:
- `Shift + Enter` → run cell and go to next cell (most common)
- `Ctrl + Enter` → run cell and stay

Cell control:
- `A` → Insert above
- `B` → Insert below
- `D, D` → Delete
- `M` → Markdown
- `Y` → Code

Modes:
- `Esc` → Command mode
- `Enter` or `Mouse L btn double-click` → Edit mode
